```python
import h2o 
from h2o.automl import H2OAutoML
```

```python
h2o.init()
```

```python
df = h2o.import_file('insurance.csv')
```

```python
df
```

```python
#separando o modelo o target
df =  df.drop('C1')
```

```python
#separando em treino e teste
treino, teste =  df.split_frame(ratios=[0.7], seed = 0)
```

```python
models_automl = H2OAutoML(max_runtime_secs= 180, sort_metric='AUTO')
models_automl.train(y="Accident", training_frame=treino)
```

```python
#conferindo modelos gerados
ranking = models_automl.leaderboard
ranking = ranking.as_data_frame()
ranking
```

```python
prever = models_automl.leader.predict(teste)
prever
```
